```python
# HyDE = Hypothetical Document Embeddings
# Instead of directly searching your documents using the user's question, 
# first ask an LLM to create a hypothetical answer/document, and then use that hypothetical answer for the search.

documents = [
    "RAG combines information retrieval with large language models.",
    "RAG can reduce hallucinations by providing relevant external information.",
    "Vector databases store embeddings and allow semantic search."
]

def generate_hypothetical_answer(question):
    # Normally an LLM would do this
    return "RAG reduces hallucinations by retrieving relevant information from external sources."

def retrieve(question, use_hyde=False):
    if use_hyde:
        search_text = generate_hypothetical_answer(question)
    else:
        search_text = question
    print("Text used for search:")
    print(search_text)

    # Pretend this is our vector search
    for document in documents:
        if "hallucination" in document.lower():
            print("\nRetrieved document:")
            print(document)

user_question = "How can RAG reduce hallucinations?"
retrieve(user_question, use_hyde=True)
```
