```python
# Search small → identify the exact relevant chunk → go back to its parent → give the LLM the larger context.
# Parent documents
parents = {
    "python": "Python is a programming language. It is easy to learn and widely used.",
    "java": "Java is a programming language. It is commonly used for enterprise applications."
}

# Small chunks that belong to a parent
chunks = {
    "chunk1": {
        "text": "Python is easy to learn",
        "parent": "python"
    },
    "chunk2": {
        "text": "Python is widely used",
        "parent": "python"
    },
    "chunk3": {
        "text": "Java is used for enterprise applications",
        "parent": "java"
    }
}


# Imagine our search found chunk1
matched_chunk = "chunk1"
print("Matched chunk:")
print(chunks[matched_chunk]["text"])


# Find the parent
parent_id = chunks[matched_chunk]["parent"]
print("\nParent document:")
print(parents[parent_id])
```
