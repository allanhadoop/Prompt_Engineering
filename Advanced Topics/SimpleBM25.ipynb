```python
from collections import Counter

class BM25:
    def __init__(self):                       # 1. CREATE BM25 Initalization
        self.documents = []
        self.word_counts = []

    def index(self, documents):               # 2. INDEX → PREPARE DOCUMENTS
        self.documents = documents
        for doc in documents:                         # Count words in every document
            words = doc.lower().split()
            self.word_counts.append(Counter(words))
        print(f"Indexed {len(documents)} documents")
  
    def score(self, query, doc_number):               # 3. SCORE → CALCULATE RELEVANCE
        query_words = query.lower().split()
        document_words = self.word_counts[doc_number]
        score = 0
        for word in query_words:
            score += document_words.get(word, 0)      # Give 1 point for every query word found
        return score
    
    def search(self, query):                          # 4. SEARCH → FIND RELEVANT DOCUMENTS
        results = []
        for i in range(len(self.documents)):          # Score every document
            score = self.score(query, i)
            if score > 0:
                results.append((score, self.documents[i]))
        results.sort(reverse=True)
        return results

documents = [
    "Python is easy to learn",
    "Python is used for artificial intelligence",
    "Java is a programming language",
    "Artificial intelligence uses Python"
]
bm25 = BM25()
bm25.index(documents)
results = bm25.search("Python artificial intelligence")

for score, document in results:
    print(score, "->", document)
```
