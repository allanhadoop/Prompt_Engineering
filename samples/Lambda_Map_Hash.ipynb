```python
num = [4,6,7,8]

add_one = lambda x: x+1
result = list(map(add_one,num))

result2 = list(map(lambda x: x * 2, num))

print(result)
print(result2)
```

```python
num = [3,5,7,0,2]

# print([num[i] for i in range(len(num)) if num[i] > 0] )
print([(lambda x: x + 1 ) (x) for x in range(len(num))])      # lambda x: x * 2 -- This creates a function and ...(x) calls that function with x  
# In above range(len(num)) -- So x is the index, not the value in the list.

print([(lambda x: x + 1 ) (x) for x in num])   # Here it goes through the values.
```
