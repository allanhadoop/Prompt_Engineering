```python
#------------Dictionary and Nested Dictionary within list-----------

patient = [
    {"patient id": 40, "diagnosis": "cancer"},
    {"patient id": 50, "diagnosis": "heart failure"},
    {"patient id": 60, "diagnosis": "liver fail"}, {        # <<--------Nexted dicionary within list
    "action item" : {                                 
        "visit on" : "15th Aug 2013",
        "proof" : "aadhar card"
    }}
]
print(patient[3]['action item']['proof']) 
print(patient[0]['diagnosis'])                         #<<<--- Value of dictionary at index 0 

for i in patient:
    print(i)

print(i['patient id'] for i in patient)                # <<-- This is a generator so it wont print values, there is no [] of list  
print([i.get('patient id') for i in patient])          #<<-- Now we have [] , also since we donot have patient id for action item block , we use .get and select only dic lines that has patient id
```

```python
#---sorted with passing a function as an argument and when to use .items()

patients = [
    {"role id": 490, "name" : "Jack"},       #  Dictionary 1 → {"role id": 490, "name": "Jack"}
    {"role id": 450, "name" : "Sam"},        #  Dictionary 2 → {"role id": 450, "name": "Sam"}
    {"role id": 400, "name" : "Lisa"},       #  Dictionary 3 → {"role id": 400, "name": "Lisa"}
    {"role id": 480, "name" : "Kavin"}       #  Dictionary 4 → {"role id": 480, "name" : "Kavin"}
]
def key_function(patients):
    return patients["role id"]

sorted_patients = sorted(
    patients,                        #--.items() is used when you are working with one dictionary (not many dictionaries like above) and want to get its key + value pairs.
#   key = lambda x: x['role id'],                      # <<<-- sorted() expect key to be a function like Lambda or def and sorted calls that function
    key = key_function,                                # << -- here there is no argument for key_function because sorted makes the call to function
    reverse = True
)

print(sorted_patients[:])
```

```python
#-----------JSON Transformation--------
from datetime import date
patients = [
    {"name" : "Jack", "DOB" : 1975},
    {"name" : "Sam", "DOB" : 1978},
]
current_year = date.today().year

updated_patients = [      # <<-------Below is dictionary structure which is being looped 
    {"patient name" : p["name"],
     "age" : current_year - p["DOB"]}        # <<<--------- Write transformation in line for all patients by looping below
    for p in patients                        # <<<--------  Loop here and transform above data like age.
]

print(updated_patients)


```

```python
#-------------JSON Schema ------------------
from jsonschema import validate 
patientdata = { 
    "name" : "Jack",
    "age"  : 37
}

patientschema = {
    "type" : "object",
    "properties" : {
        "name" : {"type" : "string"},
        "age"  : {"type" : "integer"} 
    },
"required" : ["name", "age"]
}

validate(instance=patientdata, schema = patientschema)        # <<------Validate the patient json using schema
print("Patient data has Valid JSON ")
```
