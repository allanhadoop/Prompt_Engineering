```python
#----Create a dynamic video content with AI 

import asyncio
from openai import openai
client = AsyncOpenAI() 

async def main() -> None:
    video = await client.videos.create_all_poll(
        model = "sora-2",
        prompt = "A video of a cat on a boat",
    )

    if video.status = "completed":
        print("Video completed:", video)
    else:
        print("Video creation failed": video.status)
asyncio.run(main())
```
