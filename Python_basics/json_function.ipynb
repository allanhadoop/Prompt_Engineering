```python
import json
from numpy import concat
json1 = {
    "name" : "adam",
    "id" : 1
}

def json_processing(jsoninput) -> json :
    result = []
    id = jsoninput["id"] + 1
    result = {
        "name" : f"({jsoninput["name"]}{id})" ,
        "id" : id 
    } 
    return result

json_processing(json1)
```
