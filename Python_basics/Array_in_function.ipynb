```python
import numpy as np

documents = [
    ("Oncology Doctor", np.array([1.0,0.0])),
    ("Cardiology treatment", np.array([0.3,0.2])),
    ("Heart failure", np.array([0.1,0.2]))
]

def fun(query_embedding: np.array, topk:int = 3) -> list[tuple[str, float]]:  # <<-------Passing entire array in argument
    results = []
    
    for document, embedding in documents:
        score = np.dot(query_embedding, embedding)         # <<-----Highest the dot product .. high sorted
        results.append((document, score))

    results.sort(key= lambda x:x[1], reverse=True)         # <<-------Sort results using lambda
    return results[:topk]

query = np.array([1.0,0.4])
results = fun(query,2)
print(results)
```
