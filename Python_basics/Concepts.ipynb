```python
# ---------passing one object into another class.
class first:
    def __init__(self, age:int):
        self.age = age


class second:
    def __init__(self, o1):
        self.age = o1.age + 10


o1 = first(20)
o2 = second(o1)
print("first class age", o1.age)
print("second class age", o2.age)
```

```python
# ---------Inheritance  - One class gets features from another

class father_animal_class:        # we need __init__ only for setup. You need it when you want to set values when an object is created
    def speak(self,s : str):
        self.s = s
        print(self.s)

class child_dog(father_animal_class):         #<< ASssign father class to child class and access all its methods
    def bark(self):    
        print("barking....")

d = child_dog()
d.speak("dog is animal and about to speak")
d.bark()
```

```python
# ------Composition----One class uses another class's object.
class Engine:
    def start(self):
        print("Engine started...")


class car:
    def __init__(self):
        self.engine = Engine()    # <<<<----   Create an object Engine() of the Engine class and store it in self.engine.
        self.engine.start()

c = car()
 
"""
Engine       → class
Engine()     → object created from that class
"""
```

```python
#----------------Encapsulation -- means controlling access to internal data.

class bankdetails:
    def __init__(self,balance):
        self.__balance = balance

    def get_balance(self):
        return self.__balance


account = bankdetails(1000)       
print(account.get_balance())        # << -- Access allowed only via method
print("--------------------")
print(account.self.__balance)       # <<-- Direct access not allowed - 'bankdetails' object has no attribute '__balance'
```

```python
# --- Polymorphism --------Same method name(not class), different behavior.

class one:
    def speak(self):            # <<-----------same method name
        print("bark")

class two:
    def speak(self):             # <<-----------same method name
        print("mowwww")

list_of_animals = [one(), two()]       # <<< ------list of classes

for animal in list_of_animals:
    animal.speak()
```

```python
#-----------Abstract Classes -- "Any class that inherits from Animal class (not its object) MUST create its own sound() method."
# Abstract class = a parent class that defines rules that child classes must follow.
from abc import ABC, abstractmethod

class animal(ABC):                     # << ----- this ABC and @abstractmethod is important 
    @abstractmethod                    # <<-------------------
    def sound(self):
        pass                           # <<-------------------

class dog(animal):                     # <<-------------------
    def sound(self):            
        print("barking....")

class cat(animal):                    # <<-------------------
    def sound(self):
        print("mowwww.....")

class cow(animal):
    pass

D = dog()
D.sound()

C = cat()
C.sound()

CW = cow()             # <<< This will give error cow should implement sound() method 
```

```python
#----------------Class Method-- normal method = works with one particular object but Class method works with the User class itself

class User:

    count = 0
    def __init__(self,name):
        self.name = name
    #   self.count = 1             # -- self means the individual object. so this wont work 
        User.count = 1             # -- This is at class level count
        User.surname = name        # the last time __init__() ran, it set User.surname = "Mary".

    def show_names(self):
        return self.name

    @classmethod
    def total_users(cls):
        cls.count += 10
        print(cls.count)
        print(cls.surname)

u1 = User("John")
u2 = User("Mary")

print(u1.show_names())                    #<<---------Normal method call using u1 as object

User.total_users()                #<<---------Class method - so call is calling method directly
```

```python
#----------Static Method - A static method is a function inside a class that does not need the object (self) to call itself

class calculator:

    def normal_method(self,a,b):
        sum = a + b
        return sum

    @staticmethod
    def stat_method(a,b):
        return a * b

obj = calculator()
print("normal method sum :" , obj.normal_method(4,5))

print("static method multiplier: " , calculator.stat_method(4,5))
```

```python
#---------Property -= A normal attribute stores a value directly. A property lets you put logic/validation in between 
from multiprocessing.sharedctypes import Value
# when someone reads or changes the value.

"""
Normal:
p.age → value

Property:
p.age → method → value

@property = read
@age.setter = write/change
"""
class person:
    def __init__(self, age):
        self.age = age
    @property
    def age_management(self):
        self.age += 10
        return self.age

    @age_management.setter
    def age_management(self,value:int):
        if value < 0 :
            raise ValueError("Agen cannot be negative")
        self.age = value

p = person(20) 
print("Normal class init value:", p.age)
print("Invoke property mgt method:", p.age_management)                        # << -- Normal way to read and print value of age from init and then read_age

p.age_management = 90                                          # <<--- Here we are changing or assigning new value to age so @setter section will get called.
print("With property change value using setter:", p.age_management)
```

```python
#-----------Dataclass - Upgraded way to use init assignments 
#Instead of writing below - 
class User:
    def __init__(self, name, age, country):
        self.name = name
        self.age = age

# we can use below - 
from dataclasses import dataclass    # << -----------------
@dataclass                           # << -----------------
class User:
    name: str                        # << -----------------
    age: int                         # << -----------------
user = User("John", 30)
print(user)

# Python automatically creates useful things such as:
# __init__()
# __repr__()
# __eq__()
```
