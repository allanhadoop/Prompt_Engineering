```python
# ---------passing one object into another class.
class first:
    def __init__(self, age:int):
        self.age = age


class second:
    def __init__(self, o1):
        self.age = o1.age + 10


o1 = first(20)
o2 = second(o1)
print("first class age", o1.age)
print("second class age", o2.age)
```

```python
# ---------Inheritance  - One class gets features from another

class father_animal_class:        # we need __init__ only for setup. You need it when you want to set values when an object is created
    def speak(self,s : str):
        self.s = s
        print(self.s)

class child_dog(father_animal_class):         #<< ASssign father class to child class and access all its methods
    def bark(self):    
        print("barking....")

d = child_dog()
d.speak("dog is animal and about to speak")
d.bark()
```

```python
class Engine:
    def start(self):
        print("Engine started...")


class car:
    def __init__(self):
        self.engine = Engine()    # <<<<----   Create an object Engine() of the Engine class and store it in self.engine.
        self.engine.start()

c = car()
 
"""
Engine       → class
Engine()     → object created from that class
"""
```
