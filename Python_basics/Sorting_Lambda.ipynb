```python
# sort employees by salary, highest to lowest.
salaries = {
    "Anand": 50000,
    "Rahul": 80000,
    "Priya": 60000,
    "John": 90000
}

sorted_salaries = sorted (
    salaries.items(),                               # These are all pairs . example ("Anand", 50000) with 
    key = lambda x:x[1],                            # When sorting use second value which is 50000
    reverse= True                                   # which direction- Highest → lowest
) 

for i in range(len(sorted_salaries)):
    print(sorted_salaries[i])
"""
x = ("Anand", 50000)
     ↑          ↑
    x[0]       x[1]
   name       salary
"""
```
