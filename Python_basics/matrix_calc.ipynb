```python
import numpy as np
def levenshtein_distance(s1:str, s2:str) -> int:            # calculate minimum number of edits required to transform one string into another.
    m, n = len(s1), len(s2)
    dp = np.zeros((m + 1, n + 1), dtype = int)                  # Create a matrix of size (m+1) x (n+1) .m + 1 = characters of s1 + empty string
    """
        ""  c  u  t                                           # for cut - 3 characters + 1 empty string = 4 and same for cat . So it is matrix of 4 × 4, not 3 × 3.
    +------------
    "" |  0   0  0  0
    c  |  0   0  0  0
    a  |  0   0  0  0
    t  |  0   0  0  0
    Bwlow code is used to create a starting values . "" - It's a starting point that lets the algorithm correctly handle:
        ""  c  u  t
        +------------
    ""  |  0  1  2  3
    c   |  1
    a   |  2
    t   |  3

    """
    for i in range(m + 1):
        dp[i][0] = i                                # delete all chars from s1
    for j in range(n + 1):
        dp[0][j] = j                                # insert all chars into s1

#levenshtein_distance("cat", "cut")
    # Fill the matrix
    for i in range(1, m + 1):
        for j in range(1, n + 1):
            if s1[i - 1] == s2[j - 1]:
                dp[i][j] = dp[i - 1][j - 1]  # characters match, no edit needed
            else:
                dp[i][j] = 1 + min(                         # Give smallest number.3 possible ways to make the strings match.
                    dp[i - 1][j],                           # deletion - delete character and add. cat --> ct and then ct --> cut  so 2 edits
                    dp[i][j - 1],                           # insertion - It is not useful here but for example cat and cats then 1 edit for s is required
                    dp[i - 1][j - 1]                        # substitution so a ≠ u so 1 edit required    
                )                                           # We need all three because different pairs of words require different types of edits.
    """
    | Operation        | What it does       | Example      |
    | ---------------- | ------------------ | ------------ |
    | **Substitution** | Change a character | `cat → cut`  |
    | **Deletion**     | Remove a character | `cats → cat` |
    | **Insertion**    | Add a character    | `cat → cats` |

    """
    return  dp[m][n]

 
pairs = [("cat", "cat"), ("cat", "cta"), ("cat", "cats"), ("kitten", "sitting"), ("dog", "cat")]
for a, b in pairs:
    dist = levenshtein_distance(a, b)
    print(f"  '{a}' -> '{b}' = {dist} edit(s)")
 
```
