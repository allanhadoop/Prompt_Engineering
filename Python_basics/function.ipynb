```python
docs = [ "EGFR mutation treatment",
    "Lung cancer is a EGFR related disease",
    "Diabetes treatment"
    ]
q = "EGFR"
```

```python
def test_function(query:str, docs: list[str]) -> list[str]:
    final_query = query.lower().strip()
    return [doc for doc in docs if final_query in doc.lower()] 
    """     
    Above line is short form of below - 
        results = []
        for doc in docs:
            if q in doc.lower():
                results.append(doc)
        return results
    """

test_function(q, docs)
```
