```python
# Dunder Methods 
"""
__str__()	How object appears when printed	Classes
__len__()	What len(object) should return	Custom classes
__getitem__()	Make object[index] work	Custom classes
__call__()	Make object() work	Advanced classes
__repr__()	Developer-friendly representation	Classes/debugging
"""
from typing import Self
class patient:
    def __init__(self, name, age):
        self.name = name
        self.age  = age + 100
    def __str__(self):
        return f"Patient's info: {self.name}, and age : {self.age}"

def main():                                 # A function named main that you must define yourself
    output = patient("adam", 34)
    print(output)

if __name__ == "__main__":                  #"If this file is being run directly, start the main program."
    main()
```

```python
class Patient:

    def __init__(self):
        self.medicines = ["Aspirin", "Paracetamol", "Ibuprofen"]
    def __len__(self):                             # __len__() = "If someone asks how big/long my object is, return this."
        return len(self.medicines)


patient1 = Patient()
print(len(patient1))
```

```python
class Patient:

    def __init__(self):
        self.medicines = ["Aspirin", "Paracetamol", "Ibuprofen"]
    def __getitem__(self, index):                                # __getitem__() = "If someone uses [ ] on my object, tell Python what to return."
        return self.medicines[index]

patient1 = Patient()
print(patient1[0])
print(patient1[1])
```

```python
class Patient:

    def __init__(self, name):
        self.name = name

    def __call__(self):
        print(f"Hello {self.name}")              # __call__() → Make an object behave like a function

patient1 = Patient("Adam")
patient1()
```

```python
class Patient:

    def __init__(self, name, age):
        self.name = name
        self.age = age
    def __str__(self):
        return f"Patient: {self.name}"
    def __repr__(self):                                            # __repr__() → Developer/debugging representation. it is similar to __str__
        return f"Patient(name='{self.name}', age={self.age})"


patient1 = Patient("Adam", 34)

print(patient1)
print(repr(patient1))
```
