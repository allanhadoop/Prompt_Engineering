```python
from numpy import float32
import numpy as np 
embedding = [0.25, 0.81, -0.32, 0.44]


def embedding_to_byte(embed_input: list[float]) -> bytes:
    return np.array(embed_input, dtype=np.float32).tobytes()

def from_byte_to_embedding(byte_input:bytes) -> np.ndarray:        #<<< why we use np.ndarray here instead of float32? 
                                                                        # np.ndarray = the BOX
                                                                        # np.float32  = the TYPE OF ITEMS inside the box
    return np.frombuffer(byte_input, dtype=float32)

blob = embedding_to_byte(embedding)
print("bytes: ", blob)
print("---------------------------------------")

 
embed = from_byte_to_embedding(blob)
print(embed)
```
