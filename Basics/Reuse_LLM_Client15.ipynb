```python
# ----Reuse LLM Client for new user query, no need to recreate same client over and over again   

import os, sys
sys.path.append(r"C:\Users\allan\projects\Prompt_Engineering")
from llm_config import ollama, MODEL_OLLAMA, API_KEY,OLLAMA_BASE_URL
class llm_Response:
    def __init__(self):
        self.client = ollama
    def ask(self, question):
        messages = [{"role":"system", "content": "Answer the question in only single concise line"},
                    {"role" : "user", "content" : question}]
        response = self.client.chat.completions.create(
            model = MODEL_OLLAMA,
            messages = messages
        )
        return response.choices[0].message.content

obj = llm_Response()
print(obj.ask("What is python"))
print(obj.ask("what is AI"))
print(obj.ask("what is NodeJS"))
```
