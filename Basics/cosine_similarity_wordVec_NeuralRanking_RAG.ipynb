```python
"""
Cosine similarity is commonly used in AI/embeddings. “Sine similarity” is not a standard similarity measure in the same way.
Sine and cosine are mathematical functions used to describe an angle. Cosine is used to measure similarity between vectors.


# Suppose an embedding model converts two sentences into numbers:
sentence_A = np.array([0.9, 0.8, 0.1])
sentence_B = np.array([0.8, 0.7, 0.2])

We dont understand anything from above numbers so we treat the array as a vector/arrow then calculate cosine similarity

              ANGLE
                ↓
        ┌───────┴───────┐
        ↓               ↓
      COSINE           SINE
        ↓               ↓
  direction/angle    component
  relationship
        ↓
  used for cosine
  similarity

AI/RAG model -----------
Text
 ↓
Embedding
 ↓
Array of numbers
 ↓
Vector
 ↓
Compare two vectors
 ↓
Dot product + vector lengths
 ↓
Cosine similarity
 ↓
"How similar are these two?"


"""
```

```python
import numpy as np
A = np.array([0.9, 0.8, 0.1])
B = np.array([0.8, 0.7, 0.2])

similarity = np.dot(A,B) / (np.linalg.norm(A) * np.linalg.norm(B))        #"Compare the direction of vector A with vector B, while accounting for their lengths. linalg means linear algebra"

# np.linalg(A) -- This means how long is this vector? Give me the size/length of this vector
print(similarity)    # this will be 0.99..........That means the two vectors point in very similar directions. Cosine similarity range from 1 to -1
```
