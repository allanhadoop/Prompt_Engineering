```python
#----Unpack variables/arguments-----For numeric use *args--------
def fun(*args):
    return sum([x for x in args])
print(fun(3,15,2))

#----Unpack variables/arguments-----For named key value use **kwargs--------
def customer(**kwargs):
    print(kwargs)
customer(name="John",age=34,location= 'paris')

# -----Unpack all 

def all(*args, **kwargs):
    print(args, kwargs)
all(3,5,name="sam",profession='cardiologist')


```
