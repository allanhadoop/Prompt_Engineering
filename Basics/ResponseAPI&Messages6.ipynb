```python
# Connect multiple responses using response.id-----------------
import ollama
from openai import OpenAI
import os
from dotenv import load_dotenv
import openai

from llm_config import ollama, MODEL

response = ollama.responses.create( 
        model = MODEL, 
        input = "Which country has the coldest weather?"
)
print(response.output_text)
second_response = ollama.responses.create( 
        model = MODEL, 
        previous_response_id = response.id,
        input = [{"role":"user", "content" : "explain why it is the coldest"}]
)
print("Second response : " , second_response.output_text)
```
