```python
num = [5,4,2,6,7,9]
squares = {x: x * x for x in num}      # << -- Curly brackets and extra x at the beginning acting as a Key
print(squares)
```

```python
names = ["john", "kate","jasmin"]
pairs = {x[::2] : x for x in names}              # x[start:stop:step] - step = 2  means take every 2 character 
print(pairs)
```
