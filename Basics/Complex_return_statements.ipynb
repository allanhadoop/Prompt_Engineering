```python
## ---Complex return statements --------------
num = [4,5,10,0,-2,4,9]
def sort(all:int):
    return [i for i in all if i > 4]

sort(num)
```

```python
#----another example ----------
scored = [
    ("Document A", 5),
    ("Document B", 0),
    ("Document C", 3),
    ("Document D", -1)
]

top4 = 4
def returnsometing(all):
    return [i for i in all[:top4] if i[1] > -2]

print(returnsometing(scored))
  
```
