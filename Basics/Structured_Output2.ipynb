```python
## Output structure using Pydantic Basemodel 

from openai import openai
from pydantic import BaseModel

client = openai()
class CalendarEvent(BaseModel):
    name: str
    date: str
    participants: list[str]

response = client.responses.parse(
        model = "gpt-4o-2024-08-06"
        input = [
            {"role" : "system", "content" : "Extract the event info"},
            {"role": "user", "content" : "Alice and Bob are attending a scient event on Monday",},
        ],
            text_format = CalendarEvent,
)
event = response.output_parsed
```

```python
## Image ANalysis 

from openai import openai
client = openai()
response = client.responses.create(
    model = "gpt-4.1-mini",
    input = [{
        "role" : "user",
        "content" : [
            { "type" : "input_text", "text" : "what is in this image?"},
            {
                "type" : "input_image",
                "image_url" : "https://th-i.thgim.com/public/incoming/534dkc/article71443778.ece/alternates/LANDSCAPE_1200/PTI09_08_2026_000507A.jpg"
            },
        ],
    }],
)
print(response.output_text)
```
