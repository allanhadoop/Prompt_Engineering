```python
# Text → Embedding/vector → Chroma vector database → Search
"""
-----------Track 1 --------Without LangChain----- 
    Python
    ↓
    Sentence Transformers
    ↓
    all-MiniLM-L6-v2
    ↓
    vectors
    ↓
    Chroma directly
-----------Track 2-----------With LangChain----
    Python
    ↓
    LangChain
    ↓
    HuggingFace Embeddings → MiniLM → vectors
    ↓
    LangChain Chroma → Chroma database
"""
```

```python
# Track 2 - Langchain 
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma

# 1. Some text documents
documents = [
    "Python is a programming language.",
    "AI means Artificial Intelligence.",
    "Chroma is a vector database.",
    "Embeddings convert text into numbers."
]
# 2. Create the embedding model
embeddings = HuggingFaceEmbeddings(
    model_name="all-MiniLM-L6-v2"
)

# 3. Create Chroma database and convert text into vectors
vectorstore = Chroma.from_texts(
    texts=documents,
    embedding=embeddings,
    persist_directory="./chroma_db"
)

print("Vectors stored in Chroma!")

# 4. Check how many vectors we have
collection = vectorstore._collection

count = collection.count()

print("Number of vectors:", count)

# 5. Get one vector
sample_embedding = collection.get(
    limit=1,
    include=["embeddings"]
)["embeddings"][0]

print("Vector dimensions:", len(sample_embedding))
```
