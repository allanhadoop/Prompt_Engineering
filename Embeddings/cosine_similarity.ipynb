```python
# Basics of matrix 
"""
(6,)                    → 1 dimension, 6 items
(2, 3)                  → 2 dimensions, 2 rows × 3 columns
(2, 3, 4)               → 3 dimensions, 2 × 3 × 4
reshape(2, 3)           → 2 rows , 3 columns 
                        2 rows
                        ↓
                        [10 20 30]   ← row 1
                        [40 50 60]   ← row 2
reshape(1, -1)          → We have 6 numbers (10,20,30,40,50,60) so now (1,-1) means 1 row and how many we need 6 to make up to 6 numbers so 
                        1 row × 6 columns
                        [[10 20 30 40 50 60]]
reshape(2,-1)           → It will be 2 rows and how many we need now out of 6 -- 3 columns to build this matrix ( 6/2 = 3)
                        [[10 20 30]
                        [40 50 60]]
reshape() is just preparing the data for cosine_similarity(). It does not improve, change, or calculate the similarity itself.
"""
```

```python
from sentence_transformers import SentenceTransformer
# Below is OpenAI embeeding model ,since it is paid, we will use SentenceTransformer open source model 
"""
def get_embedding(text):
    response = client.embeddings.create(
        model="text-embedding-3-small",                                       <<<<---------Paid embedding model
        input=text
    )
"""
def get_embedding(s1:str):
    model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")     <<<<---------Free embedding model
    return model.encode(s1)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

s100 = "I love machine learning"
s200 = "I enjoy studying artificial intelligence"
s300 = "I love football and basketball"

def find_similarity(s1,s2):
    e1 = get_embedding(s1)
    e2 = get_embedding(s2)
    print("e1", e1)                       # e1 is one vector
    print("e2", e2)                       # e2 is another vector
    similarity = cosine_similarity([e1],[e2])[0][0]          # sklearn is designed to compare multiple vectors with multiple vectors.[0][0] means "give me the value in row 0, column 0."
    return similarity

print(find_similarity(s100, s300))
```
