```python
# This demonstrates  how to use the OpenAI Responses API with store=False (i.e. without persistent state storage) 
# while manually managing and counting tokens using the tiktoken package.
# 1. Sending a conversation turn using the Responses API with store=False.
# 2. Counting tokens in a list of messages using a custom function num_tokens_from_messages.

from pyexpat.errors import messages
from llm_config import ollama, MODEL
import os
import tiktoken
from getpass import getpass

"""
The following function counts the number of tokens used by a list of messages. It selects an appropriate encoding based on 
the provided model name and counts tokens in each message field. 
This is particularly useful if you need to monitor token usage to stay within your model's context window or track cost.
"""
max_token = 100
def num_tokens_from_msg(message, model="llama3.2:1b"):
    try:
        enc = tiktoken.get_encoding(model)
    except:
        print("Warning: model not found. Using o200k_base encoding.")
        enc = tiktoken.get_encoding('o200k_base')
    token_per_msg = 3
    token_per_name = 1
    num_tokens = 0
    for msg in message:
        num_tokens += token_per_msg
        for key, value in msg.items():
            num_tokens += len(enc.encode(value))              # value = India is great then enc.encoding = [12345, 374, 2294] so count tokens which is 3
#            if msg.name == "name":
#                num_tokens+= token_per_name
    num_tokens += 3                           # every reply is primed with <|start|>assistant<|message|>
    return num_tokens 
```

```python
article_headings = [
    "I. Introduction A. Definition of the 2008 Financial Crisis B. Overview of the Causes and Effects of the Crisis C. Importance of Understanding the Crisis",
    "II. Historical Background A. Brief History of the US Financial System B. The Creation of the Housing Bubble C. The Growth of the Subprime Mortgage Market",
    "III. Key Players in the Crisis A. Government Entities B. Financial Institutions C. Homeowners and Borrowers"
]
```

```python
# In this example, we use the Responses API with store=False to send conversation turns. 
# By setting store=False, the conversation is not stored automatically.
# Also by using the response.usage.input_tokens and response.usage.output_tokens attributes, 
# we can manually track the token count for each response.

system_prompt = "You are a helpful assistant for a financial news website. You are writing a series of articles about the 2008 financial crisis. You have been given a list of headings for each article. You need to write a short paragraph for each heading. You can use the headings as a starting point for your writing.\n\n"
system_prompt += "All of the subheadings:\n"

message = []
for heading in article_headings:
    system_prompt += f"{heading}\n"
message.append({"role" : "developer", "content" : system_prompt})
message.append({"role": "user","content": "Write large 5 paragraphs on this particular heading"})
```

```python
current_requested_tokens = 0 
for heading in article_headings:
    response = ollama.responses.create(
        model = MODEL,
        input = message,
        store = False
    )
    current_requested_tokens = response.usage.input_tokens
    print("Current message count", len(message))
    print("Current token count", num_tokens_from_msg(message, MODEL))
```
